### ALL in ONE

```python
# ============================================================
# Полный пример: Pipeline + ColumnTransformer
# Разные обработки для разных колонок
# Всё в одном окне, с подробными комментариями
# ============================================================

# ---------------------------
# 1. Импорты
# ---------------------------

import pandas as pd                      # таблицы
import numpy as np                       # NaN, числовые операции

from sklearn.pipeline import Pipeline    # цепочка шагов
from sklearn.compose import ColumnTransformer  # разные обработки колонок
from sklearn.impute import SimpleImputer # заполнение пропусков
from sklearn.preprocessing import OneHotEncoder, StandardScaler, FunctionTransformer
from sklearn.linear_model import LinearRegression


# ---------------------------
# 2. Исходные данные
# ---------------------------

# X — признаки
X = pd.DataFrame({
    'size': [50, 60, np.nan, 80],              # числовой, есть NaN
    'rooms': [2, 3, 3, np.nan],                # числовой, есть NaN
    'district': ['A', 'B', 'A', np.nan],       # категориальный, есть NaN
    'condition': ['good', 'bad', 'good', 'good']  # категориальный
})

# y — таргет
y = [100, 150, 160, 200]


# ---------------------------
# 3. Явно указываем колонки
# ---------------------------

num_cols = ['size', 'rooms']                 # числовые признаки
cat_cols = ['district', 'condition']         # категориальные признаки


# ---------------------------
# 4. Pipeline для числовых признаков
# ---------------------------

numeric_pipeline = Pipeline(steps=[
    # 1) Заполняем NaN средним значением колонки
    ('imputer', SimpleImputer(strategy='mean')),

    # 2) Логарифмируем значения (для скошенных распределений)
    #    np.log1p(x) = log(1 + x), безопасно для 0
    ('log', FunctionTransformer(np.log1p, feature_names_out='one-to-one')),

    # 3) Масштабируем признаки (важно для линейных моделей)
    ('scaler', StandardScaler())
])


# ---------------------------
# 5. Pipeline для категориальных признаков
# ---------------------------

categorical_pipeline = Pipeline(steps=[
    # 1) Заполняем NaN самым частым значением
    ('imputer', SimpleImputer(strategy='most_frequent')),

    # 2) One-Hot Encoding
    # handle_unknown='ignore' — чтобы не падало на новых категориях
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])


# ---------------------------
# 6. ColumnTransformer
# ---------------------------

# Говорим:
# - num_cols → numeric_pipeline
# - cat_cols → categorical_pipeline

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_pipeline, num_cols),
        ('cat', categorical_pipeline, cat_cols)
    ]
)


# ---------------------------
# 7. Финальный Pipeline с моделью
# ---------------------------

model_pipeline = Pipeline(steps=[
    # 1) Предобработка
    ('preprocess', preprocessor),

    # 2) Модель
    ('model', LinearRegression())
])


# ---------------------------
# 8. Обучение модели
# ---------------------------

# Pipeline сам:
# - fit у imputer
# - fit у scaler
# - fit у onehot
# - transform данных
# - fit у модели

model_pipeline.fit(X, y)


# ---------------------------
# 9. Предсказания
# ---------------------------

predictions = model_pipeline.predict(X)

print("Predictions:", predictions)


# ---------------------------
# 10. Что происходит внутри (логика)
# ---------------------------

# X
# ├── size, rooms
# │   ├── mean imputation
# │   ├── log1p
# │   ├── scaling
# │
# └── district, condition
#     ├── most_frequent imputation
#     ├── one-hot encoding
#         ↓
#      объединение признаков
#             ↓
#       LinearRegression


# ============================================================
# ИТОГ
# ============================================================

# ✔ разные обработки для разных колонок
# ✔ всё воспроизводимо
# ✔ нет утечек данных
# ✔ один fit / predict
# ✔ production-ready подход
#
# Следующие шаги:
# - автоопределение skewed числовых признаков
# - разные пайплайны для skewed / normal
# - замена LinearRegression на LGBM / XGBoost
# - сравнение с CatBoost
# ============================================================
```
